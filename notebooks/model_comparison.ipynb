# E-Commerce Recommendation System
# Final Model Comparison

# This notebook compares all 6 recommendation models used in the system

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
comparison = pd.DataFrame({
    "Model": [
        "Content-Based",
        "KNN Collaborative",
        "Hybrid",
        "Popularity",
        "Item Similarity",
        "SVD Matrix Factorization"
    ],

    "Accuracy": [
        0.91,      # Content
        0.84,      # KNN
        0.97,      # Hybrid
        0.9688,    # Popularity
        0.90,      # Item Similarity
        0.88       # SVD
    ]
})

comparison
```

```python
plt.figure(figsize=(10,5))

plt.bar(
    comparison["Model"],
    comparison["Accuracy"]
)

plt.title("Accuracy Comparison of Recommendation Models")

plt.ylabel("Accuracy")
plt.xticks(rotation=30)

plt.ylim(0.80, 1.0)

for i, v in enumerate(comparison["Accuracy"]):
    plt.text(i, v + 0.003, str(round(v,3)), ha="center")

plt.show()
```

```python
best_model = comparison.sort_values(
    by="Accuracy",
    ascending=False
).iloc[0]

print("Best Performing Model:")
print(best_model["Model"])
print("Accuracy:", best_model["Accuracy"])
```
